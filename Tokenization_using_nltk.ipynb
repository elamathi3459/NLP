```python
corpus = """Artificial Intelligence is transforming the healthcare industry. 
From predictive diagnostics to personalized treatment plans, AI algorithms help physicians make faster and more accurate decisions. 
Natural language processing enables better understanding of medical records, while computer vision supports radiology and pathology. 
However, ethical concerns around data privacy and model bias must be addressed to ensure responsible AI deployment.
"""
```

```python
print(corpus)
```

```python
#Tokenization
# Paragraph -> Sentences
from nltk.tokenize import sent_tokenize
import nltk
nltk.download('punkt_tab')

documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
# Tokenization
#Paragraph -> words
#Sentence -> words

from nltk.tokenize import word_tokenize

word_tokenize(corpus)
```

```python
for words in documents:
    print(word_tokenize(words))
```

```python
from nltk.tokenize import wordpunct_tokenize # word's -> [word, ', s]
for words in documents:
    print(word_tokenize(words))
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer=TreebankWordTokenizer() # Full stop will be seperate word only for last sentence.
tokenizer.tokenize(corpus)
```
