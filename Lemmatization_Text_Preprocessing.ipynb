```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')

lemmatizer= WordNetLemmatizer()
```

```python
'''
POS - Noun - n
verb - v
adjective - a
adverb - r
'''
words = ["running", "runner", "runs", "easily", "fairly", "singing", "singer", "studying", "flies", "denied", "happiness", "automation"]

for word in words:
    print(word+"->"+lemmatizer.lemmatize(word, pos= 'v'))

```
